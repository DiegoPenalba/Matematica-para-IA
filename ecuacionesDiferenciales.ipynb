# Ecuaciones diferenciales

```python
import sympy as sp

#defino la variable simbolica
t = sp.symbols('t')
#defino la funcion dependiente
y = sp.Function('y')(t)
# defino la ecuacion diferencial
ode = sp.Eq(y.diff(t) + y, t)
# hago la solucion general de la ecuacion diferencial
sol_gral = sp.dsolve(ode, y)
sp.pretty_print(sol_gral)
```

```python
# defino una condicion inicial
t0 = 0
y0 = 1
cond_inicial = {y.subs(t, t0): y0}
solucion_particular = sp.dsolve(ode, y, ics=cond_inicial)
print(f"Solucion particular con la condicion inicial y({t0})={y0}: {solucion_particular}")
```

```python
#grafico la solucion particular
import numpy as np
import matplotlib.pyplot as plt

sol_particular_func = sp.lambdify(t, solucion_particular.rhs, 'numpy')
#defino valores de t para grafico
t_vals = np.linspace(0, 10, 400)
y_vals = sol_particular_func(t_vals)

#grafico la solucion particular
plt.figure(figsize=(10, 6))
plt.plot(t_vals, y_vals, label=str(solucion_particular.rhs))
plt.xlabel('t')
plt.ylabel('y')
plt.title('Solucion Particular de la Ecuacion Diferencial')
plt.legend()
plt.grid()
plt.show()
```
