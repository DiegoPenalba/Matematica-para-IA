### Sistema de ecuaciones lineales

```python
import numpy as np
```

```python
# defino matriz de ejemplo
A = np.array([
    [2,3,-1],
    [4,1,2],
    [-2,5,3]
])

# defino el vector que tiene los resultados (terminos independientes)
B = np.array([1,5,0])
```

```python
soluciones= np.linalg.solve(A,B)
print(soluciones[0], soluciones[1], soluciones[2])
```

### Concepto de transformacion lineal

```python
# Hago transformacion de R2 a R3
# defino funcion que transforma el vector
def transform_vector(vector, transformation_matrix):
    return np.dot(transformation_matrix, vector)
```

```python
# defino vector en R2
vector_r2 = np.array([1,2])
transformation_matrix = np.array([
    [1,0],
    [0,1],
    [1,1]
])
```

```python
# aplico transforcaion lineal
transformed_vector_r3 = transform_vector(vector_r2, transformation_matrix)
print(vector_r2)
print(transformation_matrix)
print(transformed_vector_r3)
```
