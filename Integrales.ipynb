# Integrales

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
# defino la variable simbolica
x = sp.symbols('x')
# defino una funcion original
f = -x**2 + 3*x + 2
# convierto a funcion numerica
f_lambdified = sp.lambdify(x, f, 'numpy')
# creo rango de valores para x
x_vals = np.linspace(-2, 4, 400)
# evaluo la funcion original para graficar
y_vals = f_lambdified(x_vals)
```

```python
# vamos a graficar
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label=r'$f(x)= -x**2 + 3*x + 2$', color= 'blue')
plt.title('Funcion')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Calculo la integral indefinida de la funcion
funcion_indefinida = sp.integrate(f, x)
sp.pretty_print(funcion_indefinida)
```

```python
# vamos a difinir los limitas de integracion
a, b = 0, 3
# Calculo la integral definida de la funcion entre los limites a y b
area = sp.integrate(f, (x, a, b))

print(f"El area bajo la curva entre x={a} y x={b} es: {area}")
```

```python
# vamos a graficar la funcion y el area bajo la curva

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label=r'$f(x)= -x**2 + 3*x + 2$', color= 'blue')
# relleno para el area bajo la curva
x_fill = np.linspace(a, b, 100)
y_fill = f_lambdified(x_fill)
plt.fill_between(x_fill, y_fill, color='orange', alpha=0.2, label=f'Area bajo la curva: {area}')

plt.title('Funcion Y: Area bajo la curva')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.legend()
plt.grid(True)
plt.show()
```
