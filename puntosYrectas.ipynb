```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
#defino funcion para calcular distancia entre dos puntos en el plano
def distancia_plano(a, b):
    return np.sqrt((b[0]-a[0])**2 + (b[1]-a[1])**2)

#defino funcion para calcular distancia entre dos puntos en el espacio
def distancia_espacio(c,d):
    return np.sqrt((d[0]-c[0])**2 + (d[1]-c[1])**2 + (d[2]-c[2])**2)
```

```python
# defino variables para verlas en el plano
a = np.array([2,3])
b = np.array([5,7])

# defino variables para verlas en el espacio
c = np.array([1,2,3])
d = np.array([4,6,8])

#calculo las distancias
distanciaAB = distancia_plano(a,b)
distanciaCD = distancia_espacio(c,d)
print("distancia ab:", distanciaAB)
print("distancia cd:", distanciaCD)
```

```python
#para ver grafico en el plano
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.plot([a[0], b[0]], [a[1], b[1]], 'ro-')
plt.text(a[0],a[1], 'A', fontsize=12, ha='right')
plt.text(b[0],b[1], 'B', fontsize=12, ha='right')
plt.title('puntos y rectas en el plano')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()
```

```python
#grafico en el espacio
ax= plt.subplot(1,2,2, projection= '3d')
ax.plot([c[0], d[0]], [c[1], d[1]], [c[2],d[2]], 'bo-')
ax.text(c[0],c[1],c[2], 'C', fontsize=12, ha='right')
ax.text(d[0],d[1],d[2], 'D', fontsize=12, ha='right')
ax.set_title('puntos y rectas en el espacio')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.grid(True)
plt.show()
```
