# Limites y continuidad

### Importo sympy (libreria necesaria)    

```python
import sympy as sp
```

```python
# Defino variable simbolica
x = sp.symbols('x')
# Defino la funcion
f = 1/(x - 1)
# calculo el limite cuando x se aproxima a 1 desde izquierda
lim_left = sp.limit(f, x, 1, dir='-') # guion porque es negtivo
# calculo el limite cuando x se aproxima a 1 desde derecha
lim_right = sp.limit(f, x, 1, dir='+') # porque es positiva

# verifico continuidad cuando x vale 1
print(f"limite de F(x) cuando x tiende a 1 por izquierda: {lim_left}")
print(f"limite de F(x) cuando x tiende a 1 por derecha: {lim_right}")
```

#### No seria continua. Graficamos discontinuidad

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x_vals = np.linspace(0,2,400)
y_vals = [1/(val - 1) if val != 1 else None for val in x_vals] #lista por comprension
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label=r'$f(x) = \frac{1}{x-1}$')
plt.axvline(x=1, color='r', linestyle='--', label = 'x = 1')
plt.ylim(-10, 10)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title("grafico de f(x) con discontinuidad en x=1")
plt.legend()
plt.grid(True)
plt.show()
```
