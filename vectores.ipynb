```python
import numpy as np
import matplotlib.pyplot as plt
```

### Vectores en el plano

```python
# defino vectores
v1= np.array([3,4])
v2= np.array([1,2])

# operaciones
suma = v1 + v2
resta = v1 - v2
producto_escalar = np.dot(v1, v2)

print(suma, resta, producto_escalar)
```

```python
# creo la figura y eje para graficar
fig, ax = plt.subplots()
ax.quiver(0,0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='r', label='v1')
ax.quiver(0,0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b', label='v2')

#grafico suma y resta
ax.quiver(0,0, suma[0], suma[1], angles='xy', scale_units='xy', scale=1, color='g', label='suma')
ax.quiver(0,0, resta[0], resta[1], angles='xy', scale_units='xy', scale=1, color='m', label='resta')

#configuro limites y etiquetas
ax.set_xlim(-1, 7)
ax.set_ylim(-1, 7)
ax.axhline(0,color= 'black', linewidth=0.5)
ax.axvline(0,color= 'black', linewidth=0.5)
ax.grid(True, which= 'both')

plt.legend()
plt.title('Operaciones con vectores')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

### Vectores en el espacio

```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
# defino vectores
ve1 = np.array([3,4,5])
ve2 = np.array([1,2,3])

#operaciones
sumaE = ve1 + ve2
restaE = ve1 - ve2
producto_escalarE = np.dot(ve1, ve2)

print(sumaE, restaE, producto_escalarE)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.quiver(0,0,0, ve1[0], ve1[1], ve1[2], color='r', label='v1')
ax.quiver(0,0,0, ve2[0], ve2[1], ve2[2], color='b', label='v2')

#grafico suma y resta
ax.quiver(0,0,0, sumaE[0], sumaE[1], sumaE[2], color='g', label='suma')
ax.quiver(0,0,0, restaE[0], restaE[1], restaE[2], color='m', label='resta')

#configuro limites y etiquetas
ax.set_xlim(-1, 7)
ax.set_ylim(-1, 7)
ax.set_zlim(-1, 7)

plt.legend()
plt.title('Operaciones con vectores 3D')


plt.show()
```
